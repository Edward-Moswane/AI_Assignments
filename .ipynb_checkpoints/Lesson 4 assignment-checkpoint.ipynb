## Task 1: Create a DataFrame using Pandas

```python
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 28],
    'City': ['New York', 'San Francisco', 'Los Angeles', 'Chicago']})

df
```

## Task 2: Row and Column Manipulation

```python
df_no_city = df.drop('City', axis=1)

print(df_no_city)
```

## Task 3: Handling Null Values

```python
df_null = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, np.nan, 35, 28],
    'City': ['New York', 'San Francisco', np.nan, 'Chicago']})

print("DataFrame with null values:")
print(df_null)

df_null['Age'] = df_null['Age'].fillna(df_null['Age'].mean())
df_null['City'] = df_null['City'].fillna('Unknown')

print("\nDataFrame after filling null values:")
print(df_null)
```

## Task 4: GroupBy and Describe

```python
df_group = pd.DataFrame({
    'Category': ['A', 'B', 'A', 'B', 'A', 'C'],
    'Value': [10, 20, 15, 25, 30, 35]})
```

```python
grouped_description = df_group.groupby('Category')['Value'].describe()

print(grouped_description)
```

## Task 5: Concatenation and Merging

```python
df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})
df3 = pd.DataFrame({'C': [9, 10], 'D': [11, 12]})
```

```python
# Concatenate DF vertically
df_vertical = pd.concat([df1, df2], ignore_index=True)

print("Vertical_Concatenation:")
print(df_vertical)

# Merge DF horizontally
df_horizontal = pd.concat([df1, df3], axis=1)

print("\nHorizontal_Merge:")
print(df_horizontal)
```

 ## Task 6: Tuples and Sets

```python
# tuple
fruits = ('laychee', 'guava', 'papaya')

# a set
numbers_set = {1, 2, 3}

print("Original tuple:", fruits)
print("Original set:", numbers_set)

# Adding to the tuple
try:
    fruits += ('mango',)
    print("Tuple after adding:", fruits)
except TypeError as e:
    print("Cannot add to tuple:", e)

# Adding an element:
numbers_set.add(4)
print("Set after adding 4:", numbers_set)
```

## Task 8: Functions and Lambda

```python
def square(number):
    return number ** 2

print("Using regular function:", square(5))

square_lambda = lambda number: number ** 2

print("Using lambda function:", square_lambda(5))
```

## Task 9: Iterators and Generators

```python
# First 5 even numbers interator
even_numbers = [0, 2, 4, 6, 8]
even_iterator = iter(even_numbers)

print("Iterator_EvenNumbers:")
for number in even_iterator:
    print(number)

# First 5 even numbers for generator
def even_generator():
    for number in range(0, 10, 2):
        yield number

print("\nGenerator_EvenNumbers:")
for number in even_generator():
    print(number)
```

## Task 10: Map, Reduce, and Filter

```python
numbers = [1, 2, 3, 4, 5]
```

```python
# Using map to square all numbers in a list.
squared_numbers = list(map(lambda x: x ** 2, numbers))

# Using reduce to find the product of all numbers in a list
product = reduce(lambda x, y: x * y, numbers)

# Using filter to get only even numbers from a list.
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print("Squared_Numbers:", squared_numbers)
print("Product_of_Numbers:", product)
print("Even_Numbers:", even_numbers)
```

## Task 11: Object-Oriented Programming - Creating a Class

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)


rectangle = Rectangle(10, 5)

print("Length:", rectangle.length)
print("Width:", rectangle.width)
print("Area:", rectangle.area())
print("Perimeter:", rectangle.perimeter())
```

## Task 12: Pandas Data Analysis

```python
df_employees = pd.DataFrame({
    'Name': ['John', 'Jane', 'Bob', 'Alice', 'Charlie'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR'],
    'Salary': [55000, 65000, 70000, 60000, 58000]})
```

```python
# Average salary by department
average_salary = df_employees.groupby('Department')['Salary'].mean()

print("Average Salary by Department:")
print(average_salary)

# Employees with salary > 60000
high_salary_employees = df_employees[df_employees['Salary'] > 60000]['Name']

print("\nEmployees with Salary > $60,000:")
print(high_salary_employees)

# Bonus column 10% of salary
df_employees['Bonus'] = df_employees['Salary'] * 0.10

print("\nEmployee Inoformation with Bonus:")
print(df_employees)
```
